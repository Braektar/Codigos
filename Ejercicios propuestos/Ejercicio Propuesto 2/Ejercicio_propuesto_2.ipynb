```python
class TorreDeHanoi:
    
    def __init__(self):
        self.pilar_1 = [6, 5, 4, 3, 2, 1]
        self.pilar_2 = []
        self.pilar_3 = []
        
    def mover_disco(self, pilar_origen, pilar_destino):
        # Recuerda que debes sacar el elemento que está más arriba en el pilar de origen
        # y colocarlo en lo más alto del pilar de destino
        # Sacar el disco
        if pilar_destino == pilar_origen:
            print("Los pilares ingresados son iguales, movimiento invalido")
            return()
        if pilar_origen == 1:
            origen = self.pilar_1
        elif pilar_origen == 2:
            origen = self.pilar_2
        else:
            origen = self.pilar_3

        if pilar_destino == 1:
            destino = self.pilar_1
        elif pilar_destino == 2:
            destino = self.pilar_2
        else:
            destino = self.pilar_3

        if len(origen) == 0:
            print(f"El pilar {pilar_origen} se encuentra vacío, movimiento invalido")
            return(1)
        
        if len(destino) != 0:
            disco_origen = origen[-1]
            disco_destino = destino[-1]
            if disco_origen > disco_destino:
                print("El disco de origen es mas grande que el de destino, movimiento invalido")
                return(2)

        disco = origen.pop()
        destino.append(disco)
        print("Movimiento realizado")
        if self.comprobar_puzzle() == 1:
            print("Puzzle completado!")
        print(self)
        return()
    
    def __str__(self):
        output = ""
        # Range termina con -1 para recorrer al revés
        for i in range(5, -1, -1):
            fila = " "  # Armamos una fila a la vez, desde arriba
            # Pilar 1
            if len(self.pilar_1) > i:
                fila += str(self.pilar_1[i]) + " "
            else:
                fila += "x "
            # Pilar 2
            if len(self.pilar_2) > i:
                fila += str(self.pilar_2[i]) + " "
            else:
                fila += "x "
            # Pilar 3
            if len(self.pilar_3) > i:
                fila += str(self.pilar_3[i]) + " "
            else:
                fila += "x "
            output += fila + "\n"
        output += "=" * 7 + "\n"
        return output
    
    def comprobar_puzzle(self):
        pilar_esperado = [6, 5, 4, 3, 2, 1]
        if self.pilar_3 == pilar_esperado:
            return(1)
        else:
            return(0)
```

```python
hanoi = TorreDeHanoi()
print(hanoi)

movimientos = [
    (1,2),(1,3),(2,3),(1,2),(3,1),(3,2),(1,2),
    (1,3),(2,3),(2,1),(3,1),(2,3),(1,2),(1,3),
    (2,3),(1,2),(3,1),(3,2),(1,2),(3,1),(2,3),
    (2,1),(3,1),(3,2),(1,2),(1,3),(2,3),(1,2),
    (3,1),(3,2),(1,2),(1,3),(2,3),(2,1),(3,1),
    (2,3),(1,2),(1,3),(2,3),(2,1),(3,1),(3,2),
    (1,2),(3,1),(2,3),(2,1),(3,1),(2,3),(1,2),
    (1,3),(2,3),(1,2),(3,1),(3,2),(1,2),(1,3),
    (2,3),(2,1),(3,1),(2,3),(1,2),(1,3),(2,3),
]

for origen, destino in movimientos:
    status = hanoi.mover_disco(origen, destino)
    if status == 2:
        print("Error en movimiento", (origen, destino))

```
